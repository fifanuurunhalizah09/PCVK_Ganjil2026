```python
import numpy as np
import pandas as pd
import cv2 as cv
from google.colab.patches import cv2_imshow # for image display
from skimage import io
from skimage import transform
from PIL import Image
import matplotlib.pylab as plt
```

```python
imageurls = [
    "https://raw.githubusercontent.com/opencv/opencv/master/samples/data/fruits.jpg",
    "https://raw.githubusercontent.com/opencv/opencv/master/samples/data/butterfly.jpg",
    "https://raw.githubusercontent.com/opencv/opencv/master/samples/data/starry_night.jpg"
]

for url in imageurls:
    image = io.imread(url)                     #read image
    image = cv.resize(image, (0,0), fx=0.5, fy=0.5) #resize image to half size
    image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB) #convert color to RGB
    final_frame = cv.hconcat((image, image_2)) #concatenate image
    cv2_imshow(final_frame)                    #show image
    print('\n')
```

```python
tinggi = image_2.shape[0]
lebar = image_2.shape[1]
print("resolusi image: tinggi x lebar = ", tinggi, " x ", lebar)
cv2_imshow(image_2)
```

```python
image_3 = image_2.copy()

row = int(image_3.shape[0] / 2)

#garis horizontal di tengah image tanpa loop
#image_3[int(image.shape[0]/2),:]=[0,255,0]

#membuat garis horizontal di tengah image dengan loop
for x in range(image_3.shape[1]):        # jalan menyusuri kolom
    image_3[row, x] = [0, 255, 0]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

```python
image_vertikal = image_2.copy()

# Mencari titik tengah kolom
col = int(image_vertikal.shape[1] / 2)

# Jalan menyusuri baris untuk garis vertikal
for y in range(image_vertikal.shape[0]):
  image_vertikal[y, col] = [0, 255, 0]

final_frame = cv.hconcat((image_2, image_vertikal))
cv2_imshow(final_frame)
```

```python
image_diagonal = image_2.copy()

h = image_diagonal.shape[0]
w = image_diagonal.shape[1]

# Jalan menyusuri baris dari atas ke bawah untuk membentuk silang pas ke pojok
for y in range(h):
  x1 = int(y * (w - 1) / (h - 1))  # Kiri atas ke kanan bawah
  x2 = int((h - 1 - y) * (w - 1) / (h - 1))  # Kiri bawah ke kanan atas

  image_diagonal[y, x1] = [0, 255, 0]
  image_diagonal[y, x2] = [0, 255, 0]

final_frame = cv.hconcat((image_2, image_diagonal))
cv2_imshow(final_frame)
```

```python
image_horizontal = image_2.copy()

# Titik tengah baris dan kolom
row = int(image_horizontal.shape[0] / 2)
col = int(image_horizontal.shape[1] / 2)

# Batas panjang garis (100 piksel di tengah)
panjang = 100
start_x = int(col - panjang / 2)
end_x = int(col + panjang / 2)

# Garis horizontal warna MERAH (BGR: [0, 0, 255])
for x in range(start_x, end_x):
  image_horizontal[row, x] = [0, 0, 255]

final_frame = cv.hconcat((image_2, image_horizontal))
cv2_imshow(final_frame)
```

```python
image_kotak = image_2.copy()

h, w, _ = image_kotak.shape

# Menentukan ukuran dan posisi kotak
ukuran = 100
y_start = int(h / 2) - 50
x_start = int(w / 2) - 70

# Mewarnai area kotak menjadi putih
for y in range(y_start, y_start + ukuran):
  for x in range(x_start, x_start + ukuran):
    image_kotak[y, x] = [255, 255, 255]

final_frame = cv.hconcat((image_2, image_kotak))
cv2_imshow(final_frame)
```
