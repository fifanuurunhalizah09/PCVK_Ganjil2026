```python
# Ukuran gambar
h, w = img.shape[:2]

# Rata-rata intensitas citra awal
mean_awal = np.mean(img)

# Mengecilkan gambar
kecil = cv2.resize(img, (w // 2, h // 2))

# Membesarkan kembali ke ukuran awal
besar = cv2.resize(kecil, (w, h))

# Rata-rata intensitas setelah resize
mean_akhir = np.mean(besar)

print("Ukuran awal :", img.shape)
print("Ukuran setelah diperkecil :", kecil.shape)
print("Ukuran setelah diperbesar :", besar.shape)
print("Rata-rata intensitas awal :", mean_awal)
print("Rata-rata intensitas setelah resize :", mean_akhir)
print("Perubahan rata-rata :", mean_akhir - mean_awal)
```

```python
import cv2 as cv
import numpy as np
import matplotlib.pyplot as plt
from google.colab.patches import cv2_imshow
```

```python
img1 = cv2.imread('kebun.png')
img2 = cv2.imread('bunga.png')
```

```python
img1_rgb = cv.cvtColor(img1, cv.COLOR_BGR2RGB)
img2_rgb = cv.cvtColor(img2, cv.COLOR_BGR2RGB)
```

```python
print("Ukuran citra 1 (kebun.png):", img1_rgb.shape)
print("Ukuran citra 2 (bunga.png):", img2_rgb.shape)
```

```python
def buat_thumbnail(img, ukuran=400, warna_bg=(245, 245, 245), nama_toko="TOKO FIFA"):
    h, w = img.shape[:2]

    if w > h:
        new_w = ukuran
        new_h = int(h * ukuran / w)
    else:
        new_h = ukuran
        new_w = int(w * ukuran / h)

    # Mengubah ukuran gambar tanpa mengubah rasio
    img_resize = cv.resize(img, (new_w, new_h))

    # Membuat kanvas persegi
    kanvas = np.full(
        (ukuran, ukuran, 3),
        warna_bg,
        dtype=np.uint8
    )

    # Menentukan posisi gambar di tengah
    y_offset = (ukuran - new_h) // 2
    x_offset = (ukuran - new_w) // 2

    # Menempatkan gambar ke tengah kanvas
    kanvas[
        y_offset:y_offset + new_h,
        x_offset:x_offset + new_w
    ] = img_resize

    # Ukuran nama toko dibuat proporsional
    skala_font = ukuran / 600
    tebal = int(ukuran / 200)

    x_logo = ukuran - ukuran // 3
    y_logo = ukuran - ukuran // 20

    # Kotak logo
    cv.rectangle(
        kanvas,
        (x_logo - 10, y_logo - ukuran // 10),
        (ukuran - ukuran // 20, ukuran - ukuran // 40),
        (0, 0, 0),
        -1
    )

    # Nama toko
    cv.putText(
        kanvas,
        nama_toko,
        (x_logo, y_logo),
        cv.FONT_HERSHEY_SIMPLEX,
        skala_font,
        (255, 255, 255),
        tebal,
        cv.LINE_AA
    )

    return kanvas
```

```python
thumb1 = buat_thumbnail(
    img1_rgb,
    400,
    (245, 245, 245),
    "TOKO FIFA"
)

thumb2 = buat_thumbnail(
    img2_rgb,
    400,
    (245, 245, 245),
    "TOKO FIFA"
)
```

```python
plt.figure(figsize=(12, 6))

plt.subplot(1, 2, 1)
plt.imshow(thumb1)
plt.title(f"bunga.png\n{img1_rgb.shape[1]}x{img1_rgb.shape[0]}")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(thumb2)
plt.title(f"kebun.png\n{img2_rgb.shape[1]}x{img2_rgb.shape[0]}")
plt.axis('off')

plt.suptitle(
    'Thumbnail Persegi 1:1',
    fontsize=14,
    fontweight='bold'
)

plt.tight_layout()
plt.show()
```
